# *Department Region Analysis (Which Department Region has the highest number of employees)

```python
region=df["Department_Region"].value_counts()
print(region)
plt.title("Department Region",color="brown", fontweight="bold")
plt.xlabel("Department Region", fontweight="bold")
plt.ylabel("employee", fontweight="bold")
region.plot.bar(color="blue")

plt.show()
```

#### The HR department in Florida has the highest number of employees among all department-region combinations.

# * Status Analysis

```python
status=df["Status"].value_counts()
print(status)
status.plot.bar(color="yellow")
plt.title("Status Analysis", fontweight="bold")
plt.xlabel("Status", fontweight="bold")
plt.ylabel("Number of work",fontweight="bold")
plt.grid(color="gray")
plt.show()
```

#### The status analysis shows that the Pending status has the highest number of employees compared to Active status and Inactive statuses. 

```python
df.head()
```

# * Yearly Analysis (In which year joined the highest number of employees)

```python
df["Join_year"]=df["Join_Date"].dt.year
```

```python
yearly_join=df.groupby("Join_year").size()
print(yearly_join)
yearly_join.plot.bar()
plt.title("yearly analysis", fontweight="bold")
plt.xlabel("year", fontweight="bold")
plt.ylabel("employee", fontweight="bold")
plt.show()
```

#### In 2023, 228 employees joined the company, which was the highest number compared to other years.

# * Monthly Analysis(In 2023 , In which month joined the highest number of employee)

```python
df_2023= df[df["Join_year"] == 2023]
print(df_2023)
```

```python
monthly_sales=df_2023["Join_Date"].dt.month_name().value_counts()
print(monthly_sales)
monthly_sales.plot.bar(color="brown")
plt.xlabel("Join Date", fontweight="bold")
plt.ylabel("employee", fontweight="bold")
plt.title("monthly analysis", fontweight="bold")
plt.show()
```

### In 2023, in October month joined the 28 highest number of employee.

# * Which Department_Region has the max salary and min salary

```python
region_salary = df.groupby("Department_Region")["Salary"].mean()
print(region_salary)
print(region_salary.min())
print(region_salary.idxmin())
print(region_salary.max())
print(region_salary.idxmax())
region_salary.plot.bar(width=0.8,color="purple")
plt.xlabel("Department_Region", fontweight="bold")
plt.ylabel("Average Salary", fontweight="bold")
plt.title("Salary Analysis", fontweight="bold")

```

#### In Finance-Illinois departament_region has 75707.70 minimum salary and in Admin-Illinois department_region has 92050.89 the maximum salary.
